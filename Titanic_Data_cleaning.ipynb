```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\CLASSIC\OneDrive\Documents\jupyter x\Datasets\Titanic-Dataset.CSV", encoding='ISO-8859-1')
df.head()
```

```python
df.shape
```

```python
print("--- DATA  QUALITY REPORT ---")
print("Shape:", df.shape)
print("\n--- Nulls per colum---")
print(df.isnull().sum())
print("\n---Duplicates ---")
print(df.duplicated().sum()) 
print("\n---Data types ---")
print(df.dtypes)
```

```python
df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df.drop(columns=['Cabin'], inplace=True)

df.isnull().sum()
```

```python
print("Before:", df.shape[0])
df.drop_duplicates(inplace=True)
print("After:", df.shape[0])
print("Duplicates removed:", df.duplicated().sum())
```

```python
df['Sex'] = df['Sex'].str.lower().str.strip()
df['Embarked'] = df['Embarked'].str.strip()
df.head()
```

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
Upper = Q3 + 1.5*IQR

outliers = df[(df['Fare'] < lower) | (df['Fare'] > Upper)]
print("Outliers in Fare:", len(outliers))
print(f"Decision: Retain outliers - they are valid expensive tickets")
```

```python
df['Pclass'] = df['Pclass'].astype(str)
df.to_csv("Titanic_cleaned.csv", index=False)
print("Cleaned file saved!")
```
